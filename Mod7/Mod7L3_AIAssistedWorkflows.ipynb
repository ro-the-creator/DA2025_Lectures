- Now create a pipeline for the chosen variables, scaling or OneHotEncoding if needed, that fits the variables into an sm.OLS model.

- Be sure it is not transforming the entire dataset. Train-test split the variables before inputting into the pipeline, then fit into the model. Show the summary afterwards

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# select best numeric predictors (forward selection by adjusted R^2), then build a sklearn preprocessing pipeline
import statsmodels.api as sm

def forward_select_adjR2(df, y_col, candidate_vars):
    selected = []
    remaining = list(candidate_vars)
    current_adj = -np.inf
    while remaining:
        best_candidate = None
        best_adj = current_adj
        for cand in remaining:
            try_vars = selected + [cand]
            X_try = sm.add_constant(df[try_vars], has_constant='add')
            model = sm.OLS(df[y_col], X_try).fit()
            if model.rsquared_adj > best_adj:
                best_adj = model.rsquared_adj
                best_candidate = cand
        if best_candidate is not None:
            selected.append(best_candidate)
            remaining.remove(best_candidate)
            current_adj = best_adj
        else:
            break
    return selected

# candidates: numeric features (exclude price)
candidates = [c for c in num_cols if c != 'price']
chosen_vars = forward_select_adjR2(df, 'price', candidates)
print("Chosen variables:", chosen_vars)

# Prepare X/y and train-test split BEFORE any preprocessing (no leakage)
X = df[chosen_vars].copy()
y = df['price'].copy()
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# Build ColumnTransformer for chosen vars (scale numeric; if any chosen are categorical handle accordingly)
numeric_features = [c for c in chosen_vars if pd.api.types.is_numeric_dtype(df[c])]
cat_features = [c for c in chosen_vars if pd.api.types.is_categorical_dtype(df[c]) or df[c].dtype == object]

transformers = []
if numeric_features:
    transformers.append(('num', StandardScaler(), numeric_features))
if cat_features:
    transformers.append(('cat', OneHotEncoder(drop='first', sparse=False, handle_unknown='ignore'), cat_features))

preprocessor = ColumnTransformer(transformers=transformers, remainder='drop')

# Fit preprocessor on training data only (no transform on full dataset)
X_train_proc = preprocessor.fit_transform(X_train)

# Get feature names for transformed matrix
try:
    feat_names = preprocessor.get_feature_names_out()
except Exception:
    feat_names = []
    for name, trans, cols in preprocessor.transformers_:
        if name == 'num':
            feat_names.extend(cols)
        elif name == 'cat':
            enc = preprocessor.named_transformers_['cat']
            if hasattr(enc, 'get_feature_names_out'):
                feat_names.extend(enc.get_feature_names_out(cols))
            else:
                # fallback: build names from encoder categories_ (drop first due to drop='first')
                for i, col in enumerate(cols):
                    cats = enc.categories_[i]
                    for cat in cats[1:]:
                        feat_names.append(f"{col}_{cat}")

# Make DataFrame for statsmodels, add constant
X_train_df = pd.DataFrame(X_train_proc, columns=feat_names, index=X_train.index)
X_train_df = sm.add_constant(X_train_df, has_constant='add')

# Fit OLS on training data
ols_model = sm.OLS(y_train, X_train_df).fit()
print(ols_model.summary())
```

```python
import os
import streamlit as st
import pandas as pd
import numpy as np
import joblib
import sys
from importlib import metadata as importlib_metadata

# write_streamlit_app.py - writes app.py into target_dir (uses existing target_dir and pkl_path variables)

app_path = os.path.join(target_dir, "app.py")

app_code = r'''

# -*- coding: utf-8 -*-
# Required imports for the app
import joblib
import streamlit as st
import pandas as pd
import numpy as np
import sys
from importlib import metadata as importlib_metadata

# load artifact
pkl = "diamond_model.pkl"
payload = joblib.load(pkl)

preprocessor = payload.get("preprocessor")
# Determine ols_model safely (avoid evaluating truthiness of Series/arrays)
ols_model = payload.get("ols_model")
if ols_model is None:
    ols_model = payload.get("params")
if ols_model is None:
    raise RuntimeError("No valid model found in payload: missing 'ols_model' and 'params'.")
feat_names = payload.get("feat_names", [])
chosen_vars = payload.get("chosen_vars", [])
defaults = payload.get("defaults", {})
artifact_versions = payload.get("versions", {})

st.set_page_config(page_title="Diamond Price Model", layout="centered")

st.title("Diamond Price Predictor")

# version check
mismatches = []
if artifact_versions:
    runtime = {
        "python": ".".join(map(str, sys.version_info[:3]))
    }
    pkgs = ["pandas", "numpy", "scikit-learn", "statsmodels", "streamlit", "joblib"]
    for p in pkgs:
        try:
            runtime[p] = importlib_metadata.version(p)
        except Exception:
            runtime[p] = "unknown"
    for k, v in artifact_versions.items():
        rv = runtime.get(k)
        if rv and rv != v:
            mismatches.append(f"{k}: artifact={v} != runtime={rv}")
if mismatches:
    st.warning("Version mismatches detected:\n" + "\n".join(mismatches))

# Input controls
st.sidebar.header("Input features")
inputs = {}
for v in chosen_vars:
    d = defaults.get(v, None)
    if isinstance(d, (int, float)):
        step = 0.01 if isinstance(d, float) else 1
        inputs[v] = st.sidebar.number_input(v, value=float(d) if d is not None else 0.0, step=step, format="%.4f")
    else:
        inputs[v] = st.sidebar.text_input(v, value=str(d) if d is not None else "")

if st.sidebar.button("Predict"):
    # build input df
    X_in = pd.DataFrame([inputs], columns=chosen_vars)
    # transform
    try:
        X_trans = preprocessor.transform(X_in)
        # feature names from preprocessor if available
        try:
            trans_names = list(preprocessor.get_feature_names_out())
        except Exception:
            trans_names = [n for n in feat_names if n != "const"]
        X_df = pd.DataFrame(X_trans, columns=trans_names)
    except Exception as e:
        st.error(f"Preprocessor transform failed: {e}")
        raise

    # add const and align to model params
    X_df["const"] = 1.0
    # ensure order matches model params
    if hasattr(ols_model, "params"):
        params_idx = list(ols_model.params.index)
    elif isinstance(payload.get("params", None), (pd.Series, dict)):
        params_idx = list(payload["params"].index if isinstance(payload["params"], pd.Series) else payload["params"].keys())
    else:
        params_idx = ["const"] + list(X_df.columns.drop("const"))

    # create model input row with zeros for missing columns
    model_row = pd.Series({c: 0.0 for c in params_idx})
    for c in X_df.columns:
        if c in model_row.index:
            model_row[c] = X_df.iloc[0][c]
    model_X = pd.DataFrame([model_row])[params_idx]

    # prediction via params dot x
    if hasattr(ols_model, "params"):
        params = ols_model.params.reindex(params_idx).fillna(0).values
    else:
        params = np.array([payload.get("params", {}).get(c, 0.0) for c in params_idx], dtype=float)

    x_vec = model_X.iloc[0].values.astype(float)
    pred = float(np.dot(params, x_vec))

    # confidence interval if covariance available
    cov = payload.get("cov", None)
    ci_text = "N/A"
    if cov is not None:
        try:
            cov_np = cov.values if hasattr(cov, "values") else np.asarray(cov)
            # reorder cov to params_idx if cov is DataFrame
            if hasattr(cov, "index") and list(cov.index) != params_idx:
                cov_np = cov.reindex(index=params_idx, columns=params_idx).fillna(0).values
            se = float(np.sqrt(float(x_vec @ cov_np @ x_vec.T)))
            z = 1.96
            ci_low = pred - z * se
            ci_high = pred + z * se
            ci_text = f"{ci_low:,.2f} to {ci_high:,.2f} (95% CI)"
        except Exception as e:
            ci_text = f"error computing CI: {e}"

    st.metric("Predicted price (USD)", f"{pred:,.2f}")
    st.write("Confidence interval:", ci_text)

    st.subheader("Input snapshot")
    st.table(X_in.T)

st.sidebar.markdown("---")
st.sidebar.subheader("Model card")
st.sidebar.write("Features used:")
for f in chosen_vars:
    st.sidebar.write(f"- {f}")
st.sidebar.write("Known limits: trained on diamonds dataset; watch out for extreme out-of-distribution values and outliers. Model is linear; heteroskedasticity may exist.")
'''

# write the file
with open(app_path, "w", encoding="utf-8") as f:
    f.write(app_code)

print(f"Wrote Streamlit app to: {app_path}")
```
